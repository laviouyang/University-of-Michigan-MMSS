### Question 1
Create a Rectangle class with the following specifications:
Represents a rectangle whoses sides are parallel to the x and y axes. Specified by its upper-left corner, height, and width.

Add an area() method to return the area of the rectangle.

````python
class Point():
    name = "I am a Point class data attibute"
    def __init__(self, xval, yval):

        self.x = xval
        self.y = yval

    def __str__(self):
        return f"Point({self.x}, {self.y})"

    def distance(self, other_point):
        x_diff_square = (self.x - other_point.x) ** 2
        y_diff_square = (self.y - other_point.y) ** 2
        return (x_diff_square + y_diff_square) ** 0.5
````

````python
# YOUR CODE HERE
class Rectangle():
  def __init__(self, upper_left, height, width):
    self.ul = upper_left
    self.h = height
    self.w = width
  def area(self):
    return self.h * self.w
````

### Question 2

Add a `center()` method to your `Rectangle` class.

It should return a **`Point`** object located at the center of the rectangle. Remember that the rectangle is specified by its *upper-left* corner, so the center is half a width to the right and half a height below that corner.

Copy your `Rectangle` class from Question 1 into the cell below and add the new method, so that this cell defines a complete, working class on its own.

*Hint: your method should build and return a new `Point`, not a tuple.*

````python
# YOUR CODE HERE
class Rectangle():
  def __init__(self, upper_left, height, width):
    self.ul = upper_left
    self.h = height
    self.w = width
  def area(self):
    return self.h * self.w
  def __str__(self):
    return f"This is a {self.w} by {self.h} rectangle."
  def center(self):
    center_x = self.ul.x + self.w/2
    center_y = self.ul.y - self.h/2
    return Point(center_x, center_y)
````

````python
upper_left_corner = Point(0, 0)
my_rectangle = Rectangle(upper_left_corner, 2, 4)

c = my_rectangle.center()
print(c)                     # expected: Point(2.0, -1.0)
print(type(c))               # expected: <class '__main__.Point'>

# The center of a rectangle is equidistant from all four corners.
print(c.distance(upper_left_corner))
````

### Question 3: Inheritance

A **square** is just a rectangle whose height and width are equal, so `Square`
is a natural **child class** of `Rectangle`.

Write a class `Square` that inherits from `Rectangle` and does three things:

1. **Add data.** `__init__(self, upper_left, side)` takes only *one* size
   argument. Pass it to the parent's `__init__` as both the height and the
   width, then store `self.side`.
2. **Add behavior.** A `diagonal()` method that returns the length of the
   square's diagonal. Build the opposite (lower-right) corner as a `Point` and
   use the `Point.distance()` method you already have — do not re-derive the
   formula.
3. **Override behavior.** A `__str__()` method so that printing a `Square`
   says `Square` rather than `Rectangle`.

*Hint for step 1: call the parent constructor with*
`super().__init__(upper_left, side, side)`.

Notice what you do **not** have to write: `area()` and `center()` are inherited
from `Rectangle` and work on a `Square` for free. That is the whole point of
inheritance.

````python
# YOUR CODE HERE
class Square(Rectangle):
  def __init__(self, upper_left, side):
    super().__init__(upper_left, side, side)
    self.side = side
  def __str__(self):
    return f"This is a {self.side} by {self.side} square."
  def diagonal(self):
    lower_right = Point(self.upper_left.x + self.side, self.upper_left.y - self.side)
    return self.upper_left.distance(lower_right)
````
